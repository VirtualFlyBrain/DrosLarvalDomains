```python
from uk.ac.ebi.vfb.neo4j.flybase2neo.dbtools import get_fb_conn, dict_cursor
fb = get_fb_conn()
cur = fb.cursor()

driver_lookup = {}


for l in lines:
    # TODO: Add check for obsolete status.  

    query = "SELECT f.uniquename AS fbid, c.name as ftype, f.name as fname " \
            "FROM feature f " \
            "JOIN cvterm c on f.type_id=c.cvterm_id " \
            "WHERE f.name = 'P{GMR%s-GAL4}' " \
            "and c.name = 'transgenic_transposon' " % l
    cur.execute(query)
    dc = dict_cursor(cur)
    #probably best to check length = 1 and throw error if not.
    driver_lookup[l] = []
    for d in dc:
        driver_lookup[l].append(d['fbid'])
        
# Add a second cycle to look up in synonyms?
```
